```python
# Carga del archivo CSV a un DataFrame
import numpy as np
import pandas as pd
```

1. Creación del DataFrame

```python
ruta = "C:/medallas.csv"
df = pd.read_csv(ruta)

df
```

```python
2. Observaciones de los datos.
```

```python
nuevos_valores = {"Oro":0, 
                  "Plata":0,
                  "Bronce":0
                 }
df_rellenados = df.fillna(nuevos_valores)
df = df_rellenados
df 
```

```python
3. Top 3 paises con mas medallas 
```

```python
top3_paises = df.sort_values(by='Total', ascending = False).head(3)
top3_paises
```

4. Los países que tienen más de 10 medallas

```python
mas_diez_medallas = df[df['Total'] > 10]
mas_diez_medallas
```
